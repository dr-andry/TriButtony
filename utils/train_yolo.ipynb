```python
from ultralytics import YOLO
import albumentations as A
from albumentations.pytorch import ToTensorV2
import cv2
import numpy as np
import os
```

```python
DATA_YAML_PATH = '../dataset/data.yaml'
```

```python
model = YOLO('../models/yolo26s.pt')
```

```python
results = model.train(
    data=DATA_YAML_PATH,
    epochs=100,             # Количество эпох
    imgsz=416,              # Размер изображения (416 или 320 для ускорения на CPU)
    batch=1,                # Размер батча (строго 1 или 2 для CPU с 16GB RAM)
    
    # ==========================================
    # НАСТРОЙКА АУГМЕНТАЦИЙ (применяются on-the-fly)
    # ==========================================
    # Цветовые аугментации (имитация разного освещения и экспозиции)
    hsv_h=0.02,             # Сдвиг оттенка (Hue)
    hsv_s=0.7,              # Сдвиг насыщенности (Saturation) - сильно меняем
    hsv_v=0.4,              # Сдвиг яркости (Value) - имитирует тени и пересветы
    
    # Геометрические аугментации (сжатие, растяжение, поворот)
    # degrees=10.0,           # Случайный поворот от -10 до +10 градусов
    # translate=0.1,          # Случайный сдвиг изображения на 10%
    # scale=0.5,              # Масштабирование (сжатие/растяжение) от 0.5x до 1.5x
    # shear=5.0,              # Сдвиг углов (skew) до 5 градусов
    
    # Отражения (бутылки вертикальны, поэтому flipud=0.0)
    fliplr=0.5,             # Отражение по горизонтали (50% шанс)
    flipud=0.0,             # Отражение по вертикали (0% - бутылки не летают вверх ногами)
    
    # Продвинутые аугментации для улучшения обобщения
    mosaic=1.0,             # Мозаика (объединяет 4 изображения в 1, отлично для мелких объектов)
    mixup=0.1,              # Смешивание двух изображений (помогает с прозрачностью стекла)
    # copy_paste=0.1,         # Копирование и вставка объектов (имитирует плотные ряды бутылок)
    
    # ==========================================
    # ОПТИМИЗАЦИЯ ПОД CPU (i7 1255u, 16GB RAM)
    # ==========================================
    device='cpu',           # Принудительное использование CPU
    workers=2,              # Минимальное кол-во потоков (чтобы не перегружать RAM)
    cache=False,            # Отключаем кэширование изображений в оперативную память
    patience=20,            # Ранняя остановка, если нет улучшений 20 эпох подряд
    
    # Сохранение
    project='runs',
    name='one',
    exist_ok=True,
    save=True,
    save_period=10,         # Сохранять чекпоинт каждые 10 эпох
    
    # Оптимизатор
    optimizer='auto',
    lr0=0.01,
    lrf=0.01,
    cos_lr=True,            # Косинусное расписание скорости обучения (дает лучшую сходимость)
)

print("\n✅ Обучение завершено!")
print(f"Лучшая модель сохранена здесь: runs/one/weights/best.pt")

# 4. Быстрая проверка на валидации
best_model = YOLO('./runs/weights/best.pt')
metrics = best_model.val()
print(f"\n📊 Метрики:")
print(f"mAP50:    {metrics.box.map50:.4f}")
print(f"mAP50-95: {metrics.box.map:.4f}")
```
